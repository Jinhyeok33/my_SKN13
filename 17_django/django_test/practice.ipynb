```python
## Jupyter Lab에서 Django shell을 실행하기 위한 설정 / in ipynb
import os
import django
# 환경변수로 config/settings.py의 위치를 설정
os.environ['DJANGO_SETTINGS_MODULE'] = 'config.settings'
os.environ['DJANGO_ALLOW_ASYNC_UNSAFE'] = 'true'

django.setup()
```

```python
from polls.models import Question, Choice
# DB Manager
Question.objects    # <django.db.models.manager.Manager at 0x24a301361d0>
# Table 내용 조회
Question.objects.all()  # <QuerySet [<Question: Question object (1)>]>
```

```python
# 데이터 추가
from django.utils import timezone
# Question 객체 생성
q = Question(
    id=1,
    question_text="What is your favorite color?", 
    pub_date=timezone.now())
# 저장 -> 데이터 베이스에 추가 : migrate 한거랑 같은 효과야 
q.save()
q.id, q.question_text, q.pub_date
```

```python
# 데이터 변경
q.question_text = "What's your favorite color?"
q.save()
q.id, q.question_text, q.pub_date
```

```python
# 데이터 삭제 
# -> 객체 선택 후 delete()
Question.objects.all().delete()
```

```python
# 데이터 조회
# 상세 조회를 위해 데이터의 특정 컬럼, 속성값에 접근하여 조건을 걸 경우 "__" 를 사용
# filter 메소드
## 조건 만족하는 여러 데이터를 가져오기 위한 메소드
question_manager = Question.objects
question_manager.filter(id=1)
question_manager.filter(question_text__startswith="What")[0:2]

# get 메소드
## 정확히 일치하는 하나의 데이터만을 가져오기 위한 메소드 -> PK 이용 
from django.utils import timezone
# question_manager.get(pub_date__year=timezone.now().year)
question_manager.get(pk=1).was_published_recently()  # 자동적으로 테이블 상 PK 로 설정됨
question = question_manager.get(id=1)
type(question)     # <class 'polls.models.Question'>

# FK 타고 내려가기 Question -> Choice
choice_manager = question.choice_set        # 얘도 Manager
choice = choice_manager.create(choice_text="Black", votes=17)   # Choice 객체
```

```python
choice_manager.all()     # <QuerySet [<Choice: Red>, <Choice: Blue>, <Choice: SkyBlue>]>
choice_manager.count()   # 5
choice      # <Choice: BlueBlack>
choice.question     # <Question: What is your favorite color?>
```
