##  원하는 형태(shape) 텐서 생성

## 특정 타입의 Tensor를 직접 생성
- torch.tensor()로 생성하면서 dtype을 지정하면 아래 타입의 Tensor객체가 생성된다. 지정안하면 알아서 자동 설정
- 원하는 Type의 Tensor클래스를 이용해 직접 생성해도 된다.
- **torch.FloatTensor(자료구조)**
    - float32 타입 텐서 생성
- **torch.LongTensor(자료구조)** 
    - int64 타입 텐서생성
- 그외
    - BoolTensor(bool), CharTensor(int8), ShortTensor(int16), IntTensor(int32), DoubleTensor(float64)
    
## tensor 상태 조회
- **tensor.shape, tensor.size(\[축번호\])**
    -  tensor의 shape조회 / size의 경우 축 방향을 받는 메소드야.
- **tensor.dtype, tensor.type()**
    - tensor 원소들의 데이터타입 조회
    - dtype은 **data type**을 type()은 tensor **객체의 클래스 타입**을 반환한다.
- **tensor.ndim, tensor.dim()**  : tensor 차원
- **tensor.numel()**: 전체 차원에 대한 모든 원소 개수

## 난수를 이용한 생성

- **torch.rand(\*size)**: 0 ~ 1사이 실수로 구성된 배열을 생성. 각 값은 균등분포를 따른다.
- **torch.randn(\*size)**: 표준정규분포(평균:0, 표준편차:1)를 따르는 실수로 구성된 배열 생성
- **torch.randint(low=0, high, size)**: 지정한 범위의 정수로 구성된 배열 생성
- **torch.randperm(n)**: 0 ~ n-1 사이의 정수를 랜덤하게 섞은 값을 원소로 가지는 배열 생성

```python
#torch.manual_seed(0)  # seed 설정
torch.rand(1, 3, 5) # 100 x 3 x 5 / 0~1 실수
torch.randn(30, 3) # 30 X 3 / 표준정규분포 실수 -> 당연히 실수밖에 없어
torch.randint(1, 100, (3, 3, 6))  # rand의 정수 버전. 구간도 정해줘야해 
torch.randperm(100) # 0 ~ 99 를 섞어서 구성 -> randint의 편의성 버전으로 index 생성 등에 유용하게 사용할 수 있다.
```

```python
a = torch.arange(100)
a
# a 섞기
idx = torch.randperm(100) # 0 ~ 99 섞어서 반환. -> index
idx
b = a[idx] # tensor로 fancy indexing이 가능해 !@
### FancyIndexing의 경우 기존 텐서와 indexing 텐서를 모두 기억하기 어렵기에 새로운 Tensor를 만들어 관리하게 됨. 참조 X
```

## Tensor gpu/cpu 메모리로 옮기기

- pytorch는 데이터셋인 tensor를 cpu메모리와 gpu 메모리로 서로 옮길 수 있다.

    - 장치는 문자열로 설정한다.
        - CPU 사용: "cpu"
        - nvida GPU: "cuda"
        - Apple m1: "mps"
            - pytorch 1.12 부터 지원
- 옮기기
    - tensor 생성시 `device` 파라미터를 이용해 설정
    - `tensor.to(device)`를 이용해 설정
- 현재 실행환경에서 어떤 장비를 사용할 수 있는지 확인
    - nvidia gpu 사용가능확인

        - `torch.backends.mps.is_available()` - M1 사용가능 여부

```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# device = 'mps' if torch.backends.mps.is_available() else 'cpu'
# device = 'cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu'
device, torch.cuda.is_available()
```

- tensor객체.reshape(\*shape) / ~~view(\*shape) 이용~~

```python
a=torch.rand(12) # 0~1
a2 = a.reshape(3,4)
a3 = a.reshape((3,2,2))
a4 = a.reshape((3,2,-1))  #한 개 axis는 -1로 설정가능하고 그럼 계산해서 알아서 설정해 준다.
print(a.shape, a2.size(), a3.shape, a4.shape)
```

## dummy 축 늘리기

- ~~None을 이용 (numpy의 newaxis 대신 None을 사용한다.)~~
- unsqueeze(dim=축번호)

```python
import torch
a = torch.tensor([[10,20],[10,20]])
print(a.shape)

a1, a2 = a[None, :], a.unsqueeze(dim=0) # dim=정수만가능 / axis 변수와 비슷한 느낌이야 
print(a1.shape, a2.shape)

a3, a4 = a[:, :, None], a.unsqueeze(dim=-1)   # a[:, None] # 0축 :, 1축 None, 2번축 지정안함 -> 원래대로 그래서 (2, 1, 2) 가됨. / 이때는 :이 딱 하나의 축만을 의미 
print(a3.shape, a4.shape)

a5, a6 = a3[:,None,:,:, None],  a3.unsqueeze(dim=1)  # unsqueeze(): dim=정수만가능 (한번에 dummy축은 하나만 추가가능.)
print(a5.shape, a6.shape)
```

## dummy 축 제거
- squeeze(\[dim=축번호\]) 이용
- 당연히 squeeze! 쥐어짜는거니까 dummy 축만 지워져 

```python
t = torch.rand(3, 1, 4, 1, 5, 1)
print(t.shape)

r1 = t.squeeze()  #축을 명시 하지 않으면 모두 제거
print(r1.shape)

r2 = t.squeeze(dim=1) # 특정 axis 제거
print(r2.shape)

r3 = t.squeeze(dim=[1,3]) # 여러 axis의 dummy 축 제거
print(r3.shape)
```

# 값의 위치(index) 변경
- tensor 원소의 index의 위치를 바꾼다.
- `tensor.transpose(axis1, axis2)` 
    - 두 축의 자리만 변경 할 수 있다.
- `tensor.permute(axis1, axis2, axis3, ..)`
    - 두 개 이상의 축 자리를 변경한다.

```python
X = torch.arange(24).reshape(2, 3, 4)
print(X.shape)

y = X.transpose(1, 2) # 마찬가지로 축만을 건드리는 애들은 다 참조 관계가 형성돼 | 너무 데이터가 크잖아? 
y[0, 0] = 999
print(y.shape, y, X)

z = X.permute(2, 0, 1)
print(z.shape)
```

# autograd(자동미분)
- 자동 미분을 이용해 gradient(미분계수)를 계산하는 pytorch system.
- 머신러닝 모델의 경우 사람의 손으로 미분된 식을 넣어 경사하강법을 적용시켜 주어야 한다.
- 딥러닝 모델에서 weight와 bias tensor들(Parameter)은 backpropagation(역전파)를 이용해 gradient를 구해서 loss가 줄어드는 방향으로 update를 하게된다.
- pytorch는 이런 미분 수행을 자동으로 처리해 준다.
    - gradient(기울기)를 구한다는 것은 미분을 한다는 것을 말한다.
- tensor가 미분 가능하려면(gradient 계산 대상 변수) `requires_grad=True` 로 설정되어 있어야 한다. (default: False)
    - 그래야 처음 결과를 예측하는 Forward 과정에서 gradient를 구하기 위한 도함수를 미리 같이 계산하여 가지고 있을 수 있다.

> - **미분**
>   -  (순간) 변화율을 계산한다.
>   - $$\frac{\partial y}{\partial x}$$
>       - x에 대한 y의 변화율. 즉 x가 변하면 y는 얼마나 변할까?

```python
import torch
x = torch.tensor([1.0], dtype=torch.double, requires_grad=True) # 당연히 attribute 값이니 나중에 바꿔줘도 돼 
print(x.requires_grad, x, x.shape)
```

```python
# Tensor의 Attribute
x.data # 실제 Tensor에 넣어준 '값'
x.grad # Backward에서 계산된 Gradient '값'
x.shape, x.ndim, x.names # names는 각 차원의 고유한 이름 | 속성값으로 들어가 있음 
```

```python
# Tensor 데이터 x 를 다루는 Feature Extracting Function y, 얘도 Tensor야 
y = x ** 2
y, y.grad_fn, y.data, y.grad # requires_grad=True 인 Tensor를 기준으로 함수식이 적용될 때 자동미분이 적용된 함수 -> y에 같이 저장을 해놔
# PowBackward0 : 제곱 연산의 미분을 수행하는 "함수 미분 방식"
# MulBackward0 : 곱셈 연산의 미분을 수행 / 3*x꼴
# AddBackward0 : 여러 미분 함수들을 합침 
# 맨 뒤 숫자는 grad_fn 을 구분하는 index
```

```python
# Gradient 계산
y.backward()
x.grad
```

```python
'''정리
Tensor로 이루어진 여러 Layer는 각각의 Tensor층을 지나갈 때마다 고유한 계산식을 포함하게 되며,
이 계산식을 통해 이전 Tensor의 미분, Gradient를 계산할 수 있다.
이를 Chain Rule과 함께 활용하여 Parameter에 대한 경사하강법을 진행할 수 있는 것이다. 

연산 그래프의 경우 각 계산값과 연산자를 나타내는 노드와 이들을 연결하는 엣지로 이루어져 있다.
각각의 Tensor 값이 계산값 노드를 의미하며, 만일 어떠한 Tensor 가 다른 Tensor 를 연산식으로써 이용한다면 해당 Tensor의 grad_fn에
    그 연산자를 자동 미분하는 grad_fn 노드(연산자 노드)가 저장된다.
    그 grad_fn 노드는 사용한 Tensor 객체에 대한 정보를 참조하고 있으며, 이때 사용된 함수식을 참조한다.
마지막으로 이들의 참조 관계가 엣지를 이루며, 이 그래프는 여러 Layer에 걸쳐 계승된다. 
'''
```

```python
a = torch.tensor([2.0], requires_grad=True)
b = torch.tensor([3.2]) # r_grad=False
y = a - b + 3 # grad_fn 정의
y.backward() # grad를 계산 - a에 대해서만 계산 
a.grad, y.grad_fn # 그래프가 간단해서 자동으로 clean-up 되지 않음
```

```python
# 간단한 경사하강법 구현
input_data = torch.tensor([35.2])
output_data = torch.tensor([10.5]) 
weight = torch.tensor([0.2], requires_grad=True) # parameter
y_pred = weight * input_data * 2 # 모델 예측값 
loss = 2*(output_data - y_pred + weight)**2 # 오차계산-MSE
# losses = loss * output_data
loss.backward()
```

```python
y_pred, loss, y_pred.grad_fn, loss.grad_fn#, losses.grad_fn
weight.grad, y_pred.grad, weight, loss
### 어떤 weight, 가중치 변수의 Gradient란 입력 값에 대한 해당 가중치의 순간 변화율  |  당연히 입력값, 연산식에 따라 변화하는 값임 
```

```python
# grad_fn 은 각각에 연결된 이전 마지막 1개의 연산 노드의 주소이며, 이 연산 노드에는 연산 방법 / 참조 텐서등의 정보가 포함되어 있다.
# 자동미분 상의 메모리 최적화를 위해 동일 연산 노드를 여러 참조 텐서에 대해 여러 번 사용할 가능성이 존재한다. 
# None 값의 경우 해당 입력값이 더 이상 연산 그래프에서 활용되지 않는다는 의미를 가진다.
# 아래의 경우 AccumulatedGrad 를 통해 weight에 대한 gradient 를 누적시켜간다. 
print(loss.grad_fn)
print(loss.grad_fn.next_functions)
print(loss.grad_fn.next_functions[0][0].next_functions)
print(loss.grad_fn.next_functions[0][0].next_functions[0][0].next_functions)
print(loss.grad_fn.next_functions[0][0].next_functions[0][0].next_functions[0][0].next_functions)
print(loss.grad_fn.next_functions[0][0].next_functions[0][0].next_functions[0][0].next_functions[1][0].next_functions)
print(loss.grad_fn.next_functions[0][0].next_functions[0][0].next_functions[0][0].next_functions[1][0].next_functions[0][0].next_functions)
# print(loss.grad_fn.next_functions[0][0].next_functions[0][0].next_functions[1][0].next_functions[0][0].next_functions[0][0].next_functions[0][0].next_functions)
```

## torch.no_grad() 
- no_grad()를 with 구문에서 연산을 할 경우 with block 내에서 requires_grad=True로 설정된 변수를 이용한 계산식이 있더라도 grad_fn을 생성하지 않는다. 즉 gradient를 계산하지 않는다.
- 딥러닝 모델을 평가 하거나 추론을할 때는 gradient를 계산할 필요가 없기 때문에 no_grad 구문을 사용한다.


```python
x = torch.tensor([1.0], requires_grad=True)
with torch.no_grad():
    y = x**2
    print('결과:', y.data.numpy())
    print('grad_fn:', y.grad_fn)
```

## gradient 값 초기화
- backward() 에서 계산되어 저장된 gradient값을 초기화한다.
- 반복 학습을 할 경우 gradient가 누적되는 문제가 발생한다. 그래서 한번 학습이 끝나면 다음 학습 전에 초기화하는 작업을 해야 한다.

```python
x = torch.tensor([1.0], requires_grad=True)
print(x.grad)
y = x**2
z = 2*y**2
z.backward()
print(y.requires_grad)
print(x.grad, y.grad, z.grad)
```

```python
# x.grad = None -> 이게 초기화 방법 
t = x**3 # 같은 Tensor를 여러 번 이용할 경우 서로 다른 노드로 사용되어 구성된 연산그래프로 판단해 grad 값을 누적하게 된다.
        # 그래서 초기화가 필요한 것 
t.backward()
print(x.grad) 
```
