- 중앙집중식 버전 관리(CVCS) : 모든 디렉토리를 중앙 서버에서 가지고 개인은 파일들의 스냅샷을 가지고 작업한다.
- 분산 버전 관리(DVCS) : 각 개인이 모든 디렉토리의 복사본을 가지고 작업한다.

#### 1.3 What is Git?
**Git의 특징**
- 다른 VCS들은 델타 기반 버전 제어 방식, 즉 시간 경과에 따라 대상 파일의 변경사항만을 저장한다. 반면, Git의 경우 시간 경과에 따라 변화한 파일의 스냅샷을 저장하고 이에 대한 참조를 저장한다.
- Git은 로컬 저장소에 거의 대부분의 디렉토리가 저장되어 있으므로 서버와의 동기화 시간 소요가 필요하지 않으며 오프라인에서의 작업이 가능함을 의미한다.
- Git은 각각의 파일 또는 디렉토리 등에 대해 ***SHA-1 해시 메커니즘***을 사용해 해당 내용에 기반, 고유한 16진수 40자리 문자열로 이루어진 해시값을 부여한다. 이를 통해 각 파일의 주소 추적 및 변경 여부를 쉽게 확인할 수 있다.
- Git은 Git 데이터베이스에 데이터를 쌓기만 한다. 이러한 특징으로 인해 한 번 파일의 스냅샷을 커밋, 푸쉬한 후에는 내용을 잃기 힘들다.

**Git 데이터의 3가지 상태**
- Working directory (작업 트리) : 현재 사용자가 작업을 하는 디렉토리(경로)를 의미한다. Git repository로부터 단일 버전을 체크아웃해 사용한다. 여기에서 파일을 수정할 경우 파일의 해시값이 변경되며 이러한 파일을 ***'modified' 상태***라고 한다. 이들은 따로 저장되지 않은 상태이므로 adding 명령어(git add)를 통해 저장을 명령해야하며, 해당 명령을 받은 파일을 ***'staged' 상태***라고 한다.
- Staging Area (스테이징 영역) : staged 된 파일들이 모여있는 곳으로 일반적으로 Git repository 내에 존재하는 영역이다. 아직 저장되지는 않은 상태로 commit 명령어(git commit -m '커밋한 내용에 대한 메세지')를 통해 저장해주어야하며, 이렇게 저장된 파일을 ***'commited 상태***라고 한다.
- Git Repository (로컬 저장소) : 내 컴퓨터에 존재하는 로컬 저장소로 초기에 root directory에서 git init 명령어를 입력하는 순간 자동으로 생기는 디렉토리(폴더)이다. '.git'으로 폴더가 자동으로 생성되며 기존 버전의 데이터들과 새롭게 커밋된 데이터(변경사항들)들이 모두 저장되어 있다. 이곳에 존재하는 데이터는 ***Remote repository (원격 저장소)***와 git clone 명령어를 통해 연동시켜줄 수 있으며, 이렇게 연동된 경우, git pull / push 명령어를 통해 변경사항에 해당하는 내용을 받거나 밀어서 동기화시킬 수 있다.

### 2. Git Basics
#### 2.1 Getting a Git Repository
**Git repository 만들기**
- 해당 프로젝트를 진행하는 디렉토리에서 git init 명령어를 통해 필요한 모든 저장소 파일이 들어있는 '.git' repository를 얻을 수 있다.
- 만약, 새시작이 아니라 기존 디렉토리의 버전 관리를 하려면 해당 파일들에 대해 add와 commit을 진행시켜 주어야한다.

**Git 연동하기**
- 만일, 기존 Git 저장소의 사본을 가져오는 경우라면 복제를 받을 디렉토리에서 git clone \<url\> [프로젝트 파일명] 명령어를 통해 프로젝트에 대한 모든 파일을 모든 버전에 대해 받을 수 있다.
- 만일, 기존에 쓰던 프로젝트 폴더를 외부 Remote repo.에 복제하고 연동시키는 경우라면 git add remote \<origin\> \<url\> 명령어를 통해 Remote repo.의 이름을 저장하고 연동시킬 수 있다.

#### 2.2 Recording Changes to the Repository
**파일 상태 확인**
- git status 명령어를 통해 현재 디렉토리의 상태를 판단할 수 있다. 이 명령어는 현재 내가 있는 디렉토리의 branch(root:master)를 알려주고, 이 branch와 origin(remote repo.)/master(root branch) 의 버전 상태를 비교해준다.
- 추적되지 않은 상태의 파일(새로 생긴 파일)은 'Untracked files:', 추적된 상태의 파일(본래 있었던 내용이 변경된 파일)은 'Changes not staged for commit:' 출력을 통해 확인할 수 있다.
- git status -s(or --short) 명령어를 통해 위 내용을 간단히 확인할 수도 있다. 이 경우 Untracked - ??, Modified - M, Staged - A 가 출력된다.
- git diff 명령어를 통해 더 상세한 작업들을 진행할 수 있지만 그만 알아보자...

**파일 무시**
- 일반적으로 로그 파일등 자동으로 생성된 파일들의 경우 git에서 자동으로 추적하지 않는 파일 클래스가 존재한다.
- .gitignore 라는 파일명으로 작업하는 root directory 내에 자동 생성되며 수동 변경, 추가도 가능하다... 나중에 알아보자.

**파일 add, commit하기**
- git add 명령어를 통해 해당 디렉토리 내 모든 파일들을 재귀적으로 추적 및 스테이징 시킬 수 있다
- git commit -m '{message}' 명령어를 통해 커밋을 진행할 수 있다. 만일 git commit으로 명령어를 실행한다면 별도의 창이 떠 여기에 커밋 메세지를 적어주고 저장한 뒤 닫아주어야 한다.
- commit이 진행되면 해당 커밋에 대한 SHA-1 체크섬이 기록되며 이 행동은 스테이징 영역에 해당하는 변경사항을 스냅샷으로 기록하는 것이다.
- **SHA-1 체크섬** : 160비트의 해시값으로 해시 알고리즘을 통해 파일의 모든 데이터를 기반으로 생성된 파일에 고유한 값이다. 이는 마치 디지털 지문과도 같아서, 이를 가지고 원본 파일을 복구하는 것은 불가능하지만 파일의 동일성을 확인하는데는 문제가 없다. 더 강화된 버전인 SHA-256과 같은 버전도 존재한다.
정정) Git에는 4가지의 주요 객체가 존재하는데 이 객체의 데이터를 기반으로 체크섬이 계산이 된다.
- git -a -m '{message}' 명령어를 통해 스테이징 과정을 생략할 수 있다.

**파일 삭제**
- git rm {파일명} 명령어를 사용해서 디렉토리를 삭제하며 git의 추적을 없앨 수 있다.
- git rm --cashed {파일명} 명령어를 사용해서 해당 파일을 로컬에서 유지하며 git의 추적만을 없앨 수도 있다.
- git rm log/\\* .log 명령어를 통해 파일 확장자가 .log인 모든 파일을 제거할 수도 있다.

#### 2.3 Viewing the Commit History

- git log -p(=patch) -2 명령어를 사용하여 가장 최근 2개의 기록을 patch(변경 사항의 텍스트 기록)형태로 볼 수 있다.

#### 2.5 Working with Remotes
**원격 저장소**
- 내 로컬 저장소에 대해 여러 원격 저장소(github에서 관리하는 서버에 존재)와 연결을 시킬 수 있다.
- git remote : 현 저장소에 대해 연결된(fetch와 push를 어디로 할지) 원격 저장소의 이름을 볼 수 있다.
    - git remote add \<shortcut\> \<url\> : 연결할 원격 저장소를 추가할 수 있다.
    - -v : url 을 보는 옵션션
- git clone \<url\> : 현 저장소와 원격 저장소를 연결한다. + 내 main 브랜치로하여금 origin/main 브랜치를 추적하도록 자동설정하여 이후 명령어들에서 따로 원격 브랜치의 이름을 적어줄 필요가 없게 한다.
    - git branch -vv : 내 브랜치가 추적하는 원격 브랜치 확인
    - git branch -r : 내 저장소에서 추적하는 브랜치 확인

**Fetching**
- git에서 fetch라는 작업은 연결되어 있는 특정 원격 저장소로 이동해 해시값 비교를 통해 내 로컬 저장소에 없는 모든 커밋, 태그 등의 변경 내용을 가져온다.
- 단, 데이터를 가져오기만 할 뿐 로컬 저장소에 병합하지는 않고 저장만 해놓는다. 어디에? 나의 다른 브랜치에 !
- 로컬 저장소에는 기본적으로 master(=main) branch가 있고 원격 저장소(=origin)에도 기본적으로 master(=main) branch가 존재한다. fetching을 진행하면 자동으로 로컬 저장소에 **원격 참조 브랜치**가 origin/main 의 이름으로 생성되며 이 브랜치가 변경된 모든 사항의 해시값들을 참조한 채로 유지되는 것이다.
- 이후에 직접 병합해주어야 한다. by merge or rebase
- git fetch \<remote\> \<branch\>: 원격 저장소의 main 브랜치(or 특정 브랜치)를 그대로 따오기만 해

**Pushing**
- 내 브랜치에 있는 변경 사항을 원격 브랜치에 병합X 동기화O 시키는 과정이다. **"업데이트"** 가 목적
- 조금 더 정확히는 내가 새로 변경한 내용을 원격 브랜치에 추가로 저장해놓는다고 생각하면 되며, 이러한 이유로 원격 브랜치가 내가 알던 상태로부터 조금이라도 달라졌을 경우, 푸시는 자동적으로 거부된다. 
- 이 경우, pull을 먼저 진행해 최근 변경 사항을 내 로컬 저장소에 반영해주고 오로지 내가 만든 변경 사항(커밋)만이 원격 저장소와 다른 상태에서 푸시를 진행해주어야 한다. 만일 pull에서 오류가 날 경우 stash나 rebase등을 활용할 수 있다.
- git push \<remote\> \<branch\> : 원격 저장소의 브랜치에 내용을 푸시한다. 추적 관계일 경우 위치 지정 필요없음.
    --force : 강제로 푸시 / 그냥 강제로 내 브랜치와 동기화 시킨다고 생각하면 된다. 조심할 것.

**Pulling**
- 원격 브랜치에 있는 변경 사항을 내 브랜치에 병합시키는 것을 pulling 이라고 하며, 이 작업은 fetching(브랜치째로 가져와서) merging(병합)하는 것과 다르지 않다. **"최신 트렌드 따라가기"** 가 목적
- 원격 브랜치의 최신 사항과 내 브랜치의 변경 사항이 충돌하지 않는 이상 대부분의 경우 잘 작동한다.
- 만일 내 로컬 브랜치의 최신 커밋이 원격의 최신 커밋과 충돌한다면, pull은 충돌을 해결해준 뒤, 해결 커밋을 지정해주어야 병합이 진행될 수 있으며,
만일 커밋되지 않은 staged or modified 파일이 원격의 최신 커밋과 충돌한다면, 이를 임시저장하는 stash를 이용해 병합을 진행할 수 있다.
- git pull \<remote\> \<branch\> : pulling. 추적 관계 적용.
    - git pull --rebase \<remote\> \<branch\> : 원격의 정보를 내 브랜치에 합치는 과정을 rebase로 진행하겠다는 뜻이다. 별도의 병합 커밋을 만들지 않으며 원격 브랜치의 최신 변경 사항을 먼저 내 브랜치에 적용하고, 이후 내 최신 커밋을 적용시킨다.
    - git pull --ff-only \<remote\> \<branch\> : Fast Forward 방식, 즉 이전 pull로부터 내가 아무것도 건드리지 않았을 때 가능한 pulling
    - --no-edit : pull로 생기는 커밋 메세지 처리
    - --verbose : 더 자세한 pull 이력 표시
    - --autostash : 자동 stash 기능

### 3. Git Branching

#### 3.1 Branches in a Nutshell
##### **Git을 이루는 4가지의 주요 객체**
- Git의 핵심적인 차별점인 '스냅샷을 이용한 버전 관리'는 아래 4가지의 주요 객체에 의해 이루어진다.
- 스냅샷을 찍는다는 것은 변경 내용에 대해서는 다른 VCS처럼 모든 변경 내용을 저장하지만, 그 전 스냅샷들을 차례로 참조하며 다른 모든 프로젝트의 내용까지 저장된다는 차이점을 가진다.
1. 블롭(blob) 객체 : 오로지 파일의 '변경 내용'만을 저장하는 객체이다. 파일의 위치, 구조 등에 대해서는 X.
이들은 파일의 변경 내용에 대해 고유한 SHA-1 해시값을 가진다.
2. 트리(tree) 객체 : 폴더, 디렉토리를 나타내는 객체이다. 이 객체는 블롭 객체를 참조함으로써 폴더 내 변경 내용을 한 번에 저장할 수 있으며, 다른 트리 객체를 참조함으로써 root 디렉토리로부터 시작한 전체 디렉토리 구조를 나타낼 수 있다.
이들은 전체 디렉토리의 구조와 요소에 고유한 SHA-1 해시값을 가진다.
- 위 두 가지 객체를 통해 Git은 프로젝트의 모든 변경 사항을 모든 디렉토리 별로 나타낼 수 있다.
3. 커밋(commit) 객체 : 하나 하나의 커밋을 나타내는 객체이다. 사용자의 이름, 이메일, 커밋 메세지 등 커밋의 메타 데이터를 포함한다.
하나의 커밋 객체마다 각각 새로이 하나의 root 트리 객체를 생성하여 참조하며, 이 고유한 root 트리 객체가 기존 및 새로 만들어진 모든 트리 객체와 blob 객체(의 해시)를 참조한다.
커밋 객체는 각각이 n개의 이전 커밋 객체를 참조함으로써 Git에 시간축을 만들어준다고 볼 수 있으며, 여러 세계관을 합쳐줄 수 있는 객체이다.
이들은 커밋의 메타 데이터에 고유한 SHA-1 해시값을 가진다.
4. 태그(tag) 객체 : 중요한 커밋에 들러 붙어 태그를 남겨주는 객체이다.
이들 역시 담당한 커밋의 내용에 따라 고유한 SHA-1 해시값을 가진다.

##### **Working Directory와 Root Directory**
- Root directory는 프로젝트의 최상위 디렉토리로 .git, 로컬 저장소를 가지고 있는 저장소의 기준이 되는 디렉토리이다.
- 반면, Working directory는 내가 현재 작업하는 디렉토리로 Git에서 실시간으로 추적하는 대상이 되는 디렉토리이다.
초기에는 root dir.과 동일하게 설정되어 Git과 연동된 모든 디렉토리를 실시간으로 추적하지만 프로젝트 규모가 커짐에 따라 더 작게 설정할 수 있다.
Git에서 추적한다는 의미는 working dir.의 모든 변경 사항을 커밋 대상으로 파악하며, 브랜치를 변경할 경우 자동으로 변경되는 사항들을 말한다.

##### **브랜치(Branch)**
- 브랜치라는 것은 특별하지 않다. 하나의 브랜치는 단지 하나의 커밋 객체(의 해시값)를 가리키는 포인터일 뿐이며, 이를 통해 현재 브랜치가 프로젝트의 어느 위치를 보고 있는지 알려준다.
- 처음 Git init/clone 등의 명령어를 통해 Git을 시작하게 되면 자동적으로 master라는 이름의 브랜치가 형성되어 적용된다 / master는 특별한 브랜치가 아니라 그저 가장 처음 생성된 브랜치일 뿐이다.
- 이렇게 생기는 브랜치는 그저 커밋 객체의 해시값을 가리키는 또 다른 해시값에 불과하므로 하나의 브랜치를 만드는 것은 40바이트의 글자를 만드는 것과 동일하다. 이러한 Git의 특징 덕에 Git의 사용자들은 매우 가볍게 브랜치를 생성하고 버전을 관리할 수 있게 된다.
- HEAD : 브랜치에 붙어 있는 태그로, 여러 브랜치 중 내가 현재 보고있는 브랜치를 의미한다.
- CHECKOUT : 브랜치를 옮겨다니는 행위로, HEAD를 옮기는 행위이다. Git은 현재 HEAD 브랜치에 따라 working directory의 모든 파일을을 당시 버전으로 되돌리므로 만약 커밋되지 않은 내용이 있다면 checkout이 불가능하다. 후에 해결 방법을 따로 배운다.(주로 stash나 커밋 amend)
- LOG : 해당 브랜치의 history를 보여주는 행위로 어떻게 분기되어 나왔는지, 어떤 커밋들을 지나왔는지에 대한 정보를 보여준다.

#### 3.2 Basic Branching and Merging
##### **5가지 Merge 방법**
- 버전 관리 시스템의 특성에 따라 사용자가 현재 프로젝트의 어떤 부분에 대해 작업을 하고 있던 지 간에, Git은 사용자로 하여금 프로젝트의 어느 버전에서든 간에 돌아가 다른 작업(오류 수정, 기능 추가 등)을 하게끔 서포트한다. 
- 이렇게 하나의 프로젝트에서 나누어진 여러 브랜치를 다시 통합시키는 것이 Merge 이며 여기에는 크게 5가지 종류가 존재한다.
- 병합이라는 것은 주체가 되는 HEAD 브랜치가 다른 브랜치를 끌고 오는 행위로, 병합 당한 브랜치는 자동으로 사라지지 않는다.
1. Fast Forward(빠른 병합) : 이는 대상 브랜치(병합 당하는 애)가 병합하려는 브랜치(주체 브랜치)의 바로 뒤에서 이어진 관계이다.
이 경우, 대상 브랜치는 분기없이 현재 브랜치로부터 쭉 나아간 브랜치이므로 그저 현재 브랜치에 참조된 커밋 해시값을 대상 브랜치의 것으로 옮겨주면 된다.
전체 tree 구조 상 일렬로 이어진다.
2. Three-Way Mergy(3-way 병합) : 이는 두 브랜치가 하나의 공통 조상 커밋을 가지는 관계이다.
이 경우, 각 브랜치의 참조 커밋과 공통 조상 커밋 3개를 비교하여 새로운 병합 커밋을 만들어 두 브랜치를 다시 묶게 된다.
전체 tree 구조 상 branch가 뻗었다가 main branch의 새로운 병합 커밋을 기점으로 다시 합쳐진다.
3. Squash Merge
4. Rebase
5. Conflict Merge : 이는 두 브랜치를 묶는 과정에서 같은 내용에 대해 두 가지의 수정 사항이 있어 서로 충돌하는 관계이다.
이 경우, Git이 자동으로 처리해주지 못하며 양쪽 중 하나를 고르거나 새로이 작성하여 저장한 뒤, 스테이징과 커밋 과정을 다시금 거쳐야 한다.

#### 3.3 Branch Management
##### **브랜치 관리**
- Git은 현재 존재하는 모든 브랜치를 관리할 수 있는 도구를 제공해준다.
- 이를 통해 현재 HEAD 브랜치에 대해 다른 브랜치들이 이미 병합된 상태인지, 따라서 삭제해도 데이터의 손실이 없는지에 대한 정보를 제공해준다.

#### 3.4 Branching Workflows
##### **Branch Workflow**
- 이러한 git의 branch를 활용한 협업 방법에는 어떤 것이 있을까?
1. Long-Running Branch : git의 branch와 그 merge 과정이 가볍다는 특징을 이용한 협업 방식으로 메인이 되는 프로젝트를 천천히 merge하는 방식이다.
develop, next등의 이름의 branch를 통해 프로젝트의 진행을 공격적으로 운용하며, master(main) branch에는 안정적이고 배포가능한 코드만을 merge 시키는 것이다.
프로젝트의 규모가 커질 수록 그 깊이는 깊어져 추가적으로 proposed 등의 branch를 사이에 만들어 조율할수도 있으며, 프로젝트가 안정적일수록 main branch는 전체 흐름에서 더욱 뒤쳐져 따라가게 된다.
2. Topic Branch : git의 branch가 매우매우 가볍다는 특징을 아주 잘 이용한 협업 방식으로 특정 이슈를 위해 짧은 호흡으로 분리되어 나온 브랜치를 말한다.
iss_num 등의 이름의 branch를 통해 특정 이슈를 여러 가지 방법으로 해결해 볼 수 있으며 이들 중 최선의 방법만을 채택해 merge 시킬 수 있다.
- 여태까지의 방법들은 모두 로컬 저장소를 통해 branch를 만들고, 처리한 내용이다.
