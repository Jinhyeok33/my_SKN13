## 🎟️ 티켓의 가치 LV.??

### 🎯 추첨 이벤트에서 **가장 높은 당첨 확률** 선택하기

한 쇼핑몰에서는 여러 개의 추첨 이벤트를 진행하고 있습니다.  
각 추첨 이벤트는 `[당첨자 수, 참여 인원 수, 티켓 가치]` 형태의 정보를 가집니다.

- 당신은 **오직 하나**의 추첨에만 참여할 수 있습니다.  
- 당신이 참여하면 해당 이벤트의 **참여 인원 수가 1 증가**합니다.  
- **당첨 확률**은 다음과 같이 계산합니다.  
  $$
  \text{당첨 확률} = \frac{\text{당첨자 수}}{\text{참여 인원 수} + 1}
  $$
  (단, 값이 1을 초과하면 1로 고정합니다.)
- 당첨 확률이 **가장 높은 추첨 이벤트의 인덱스**(1부터 시작)를 출력하세요.  
- 확률이 동일하면 **더 큰 티켓 가치의 인덱스**를 선택합니다.

---

### 🔒 제한사항
- 0 ≤ 참여 인원 수 ≤ 1000  
- 1 ≤ 당첨자 수 ≤ 1000  
- 1 ≤ 티켓 가치 ≤ 1000  
- 모든 티켓 가치는 서로 다릅니다.  
- 추첨 이벤트의 개수는 1개 이상입니다.

---

### 입출력 예

| raffles | result |
|---------|--------|
| `[[100, 100, 500], [1000, 1000, 100]]` | 2 |
| `[[10, 19, 800], [20, 39, 200], [100, 199, 500]]` | 1 |
| `[[50, 1, 50], [100, 199, 100], [1, 1, 500]]` | 1 |

---

### 입출력 예 #1 설명
- **1번 추첨**: 당첨 확률 = 100/(100+1) = 0.9900990099...
- **2번 추첨**: 당첨 확률 = 1000/(1000+1) = 0.9990009990...

2번 추첨의 확률이 더 높으므로 **2번**이 정답입니다.


```python
def solution(raffles):
    answer = sorted(raffles, key=lambda x: (min(1, x[0]/(x[1]+1)), x[2]), reverse=True)
    return raffles.index(answer[0])+1
solution([[10, 9, 100], [20, 19, 150], [30, 29, 120]])
```

```python
# 테스트케이스
test_cases = [
    # 기존 3개 + 변형, 비슷한 유형 20개
    ([[100, 100, 500], [1000, 1000, 100]], 2),
    ([[10, 19, 800], [20, 39, 200], [100, 199, 500]], 1),
    ([[50, 1, 50], [100, 199, 100], [1, 1, 500]], 1),

    # 변형 유형 1: 당첨자와 참여자 비율 살짝 다르게
    ([[90, 90, 400], [800, 800, 150]], 2),
    ([[15, 28, 900], [25, 50, 250], [90, 180, 450]], 1),
    ([[45, 2, 55], [110, 210, 90], [2, 2, 510]], 1),

    # 변형 유형 2: 확률 같고 가치만 다름
    ([[10, 9, 100], [20, 19, 150], [30, 29, 120]], 2),
    ([[5, 4, 500], [10, 9, 600], [15, 14, 550]], 2),
    ([[1, 0, 1000], [2, 1, 1500], [3, 2, 1400]], 2),

    # 변형 유형 3: 확률 모두 다르고 가치도 다양
    ([[7, 8, 400], [10, 10, 450], [20, 25, 300]], 2),
    ([[12, 15, 100], [14, 20, 200], [16, 25, 150]], 1),
    ([[50, 49, 50], [60, 61, 40], [70, 68, 30]], 1),

    # 변형 유형 4: 확률 1 이상인 경우 cap 처리 테스트
    ([[10, 9, 1000], [5, 4, 1200], [2, 1, 1100]], 2),
    ([[100, 99, 2000], [90, 89, 2100], [80, 79, 1900]], 2),

    # 변형 유형 5: 1개 추첨 케이스
    ([[1000, 1000, 10000]], 1),
    ([[5, 4, 50]], 1),

    # 변형 유형 6: 확률 낮지만 가치가 매우 높은 경우
    ([[1, 100, 500], [2, 150, 1000], [3, 200, 3000]], 3),

    # 변형 유형 7: 확률 동일, 가치 동일 (첫번째 인덱스 선택)
    ([[10, 9, 100], [10, 9, 100], [10, 9, 100]], 1),

    # 변형 유형 8: 확률 동일, 가치 큰 차이
    ([[10, 9, 100], [10, 9, 200], [10, 9, 150]], 2),

    # 변형 유형 9: 확률 작고 가치 작음 (꼴찌가 답)
    ([[1, 100, 10], [1, 100, 5], [1, 100, 1]], 1),
]

# 테스트 실행
import time
e = time.time()
for i, (raffles, expected) in enumerate(test_cases, 1):
    result = solution(raffles)
    assert result == expected, f"❌ Test case {i} failed: expected {expected}, got {result}"
    print(f"✅ Test case {i} passed")
s = time.time()

print("🎉 모든 테스트 통과!", "소요 시간 : ", s-e)
```

# 🎟️ 기간 내 책 읽은 수 만큼 당첨되는 이벤트 LV.??

### 🎯 문제 설명  
여러 개의 책 읽기 이벤트가 진행됩니다.  
각 이벤트는 다음과 같은 정보를 가집니다:  
- 이벤트 기간 (`start`일부터 `end`일까지)  
- 해당 기간에 채워야 할 **최소 리뷰 수**  

모든 이벤트에 대해 **공통 최대 당첨자 수** `m`이 주어집니다.  

각 참가자는 특정 날짜에 책을 읽은 기록이 있으며, 해당 날짜가 이벤트 기간 내에 속할 경우 그 리뷰가 해당 이벤트에 반영됩니다.

---

## 📋 이벤트 당첨 규칙  
1. 각 이벤트마다 기간 내 누적 리뷰 수가 `min_reviews` 이상인 참가자를 **후보**로 판단  
2. 후보 수가 `m` 이하라면 **모두 당첨**  
3. 후보 수가 `m`를 초과하면, **최초로 `min_reviews`에 도달한 순서**대로 앞에서부터 `m`명만 당첨  

---

## 🔒 제한사항  
- `events[i]`는 `[start_day, end_day, min_reviews]` 형식입니다.  
- 공통 최대 당첨자 수: 정수 `m`  
- `reviews[i]`는 `"참가자명 날짜 권수"` 형식입니다. (예: `"arin 20 2"`)  
- 참가자 이름은 영문 소문자, 최대 10명 이하  
- 날짜는 1 이상 31 이하의 정수  
- 권수는 1 이상의 정수  
- 이벤트 개수 최대 50개, 리뷰 개수 최대 1000건  


## 📤 출력  
사전순으로 정렬된 참가자 이름에 대응하여, 각 참가자가 **총 몇 번 당첨**되었는지를 담은 1차원 정수 배열을 반환합니다.


## 🔍 입출력 예

| events                  | m  | reviews                                                                                      | result    |
|-------------------------|----|----------------------------------------------------------------------------------------------|-----------|
| `[[1,25,3], [15,30,1]]` | 2  | `["arin 1 1", "arin 20 2", "bella 15 1", "carlos 17 1", "carlos 30 1"]`                     | `[1,1,1]` |

### 예제 설명  
1. **첫 번째 이벤트** (1~25일, 최소 3권, m=2)  
   - arin: 1 + 2 = 3권 → 후보  
   - bella: 1권 → 불충족  
   - carlos: 1권 → 불충족  
   → 후보 1명 ≤ m → arin 당첨  

2. **두 번째 이벤트** (15~30일, 최소 1권, m=2)  
   - arin: 20일에 첫 충족 → 후보  
   - bella: 15일에 첫 충족 → 후보  
   - carlos: 17일에 첫 충족 → 후보  
   → 후보 3명 > m → 충족 시점 순으로 bella(15일), carlos(17일), arin(20일)  
   → 앞의 2명(bella, carlos) 당첨  

→ 최종 당첨 횟수: arin=1, bella=1, carlos=1  
→ 사전순 결과: `[1, 1, 1]`    

```python
def solution(events,m,reviews):
    name_review = {}                                # 이름:[이벤트 당 리뷰 수]
    event_cnt = [0 for _ in range(len(events))]     # 이벤트 당 누적 당첨자 수
    name_win_cnt = {}                               # 이름:당첨수 "arin":4
    name_available = {}                             # 이름:[이벤트 당 참여 가능 여부] "arin":[True, False, True, True]
    print(f"이전 reviews : {reviews}")
    reviews.sort(key=lambda x: (x.split()[1], x.split()[0]))
    print(f"정렬 후 reviews : {reviews}")

    # 이벤트 별로 딕셔너리 관리
    for review in reviews:
        name, date, cnt = review.split()
        date, cnt = int(date), int(cnt)
        # 키 생성
        if not name_review.get(name, 0):
            name_review[name] = [0 for _ in range(len(events))]
            name_win_cnt[name] = 0
            name_available[name] = [True for _ in range(len(events))]
        # 이벤트 별 당첨
        for i, (startday, endday, _) in enumerate(events):
            if startday <= date <= endday:
                name_review[name][i] += cnt
                if name_review[name][i] >= events[i][2] and event_cnt[i] < m and name_available[name][i]:
                    name_win_cnt[name] += 1
                    event_cnt[i] += 1
                    name_available[name][i] = False
    print(f"'이름':[이벤트 당 리뷰 수] -> {name_review}")
    print(f"'이름':당첨수             -> {name_win_cnt}"+"\n")

    # 이름 순으로 정렬
    names = sorted(name_review.keys())
    answer = []
    for name in names:
        answer.append(name_win_cnt[name])
    return answer
solution([[1, 31, 1]], 1, ["alice 15 1", "bob 10 2"])
solution(
        [[1, 31, 5]], 1,
        ["alice 1 1", "alice 31 4", "bob 15 2", "bob 16 3", "carol 10 5"],
    )
solution([[1, 10, 1]], 1, ["bob 1 1", "alice 1 1"])
solution([[1, 10, 3]], 2, ["alice 1 1", "alice 2 1", "alice 3 1", "alice 4 1"])
```

### 🧩 변형 가능한 단어 찾기

어떤 문자열 `base`가 주어졌을 때, `base`의 연속된 부분 문자열들을 자유롭게 조합하거나 일부 문자를 다른 문자로 대체하여 만들 수 있는 단어들이 있는지 확인하려고 합니다.

예를 들어, `base`가 `"ghi"`일 때는 `"g"`, `"h"`, `"i"`, `"gh"`, `"hi"`, `"ghi"` 와 같은 연속된 문자열 조각을 이용해 다른 단어를 만들 수 있으며, 필요시 일부 문자는 다른 문자로 대체해도 됩니다.

당신은 여러 개의 문자열 `targets` 리스트가 주어졌을 때, 각 `target`이 위 조건에 맞춰 `base`로부터 만들 수 있는지 여부를 판단해야 합니다.

- `target`의 각 문자는 `base`의 연속 부분 문자열 조합 또는 치환으로 만들 수 있어야 합니다.
- 만들 수 있는 경우 `1`, 그렇지 않은 경우 `0`을 반환하세요.

---

### 🔒 제한사항

- 1 ≤ `base`의 길이 ≤ 20
  - `base`는 알파벳 소문자로 구성됩니다.
- 1 ≤ `targets`의 길이 ≤ 100
  - 1 ≤ 각 `targets`의 원소 길이 ≤ 40
  - `targets`의 원소는 알파벳 소문자로 구성됩니다.

---

### ✏️ 입출력 예

| base | targets | result |
|------|---------|--------|
| `"ghi"` | `["gig"]` | `[1]` |
| `"kkkkk"` | `["kkkkkkkkkk", "kkkkzkkk"]` | `[0, 1]` |
| `"wxyz"` | `["wxyzwxy", "xzqem", "wy", "wzwxy", "yyzz"]` | `[1, 0, 1, 1, 1]` |


---



### 💬 입출력 예 설명

#### 예제 1

- base: `"ghi"`  
  → 가능한 부분 문자열 조합: `"g"`, `"h"`, `"i"`, `"gh"`, `"hi"`, `"ghi"`

- target: `"gig"`  
  → `"g"` + `"hi"`(-> ig로 바꿔치기)
  ✅ 만들 수 있음 → `1`

---

#### 예제 2

- base: `"kkkkk"`

- targets:
  1. `"kkkkkkkkkk"`  
     → 필요한 `k`가 10개지만, base 조각 중 5개 짜리가 하나밖에 없음.....?
     ❌ 만들 수 없음 → `0`
  2. `"kkkkzkkk"`  
     → 중간 문자 하나(`z`)만 치환하면 base로 만들 수 있음  
     ✅ 만들 수 있음 → `1`

---

#### 예제 3

- base: `"wxyz"`

- targets:
  1. `"wxyzwxy"` → base의 조각을 반복해 만들 수 있음 → `1`
  2. `"xzqem"` → `"qem"` 등은 base에 없음 → `0`
  3. `"wy"` → `"w"` + `"y"` → 가능 → `1`
  4. `"wzwxy"` → `"wz"`는 없음이지만 일부 치환으로 가능 → `1`
  5. `"yyzz"` → 일부 치환 포함해 가능 → `1`

✅ 결과: `[1, 0, 1, 1, 1]`

### 🌱 최소 생장력 이상을 만족하는 온실 그룹 조합 수 구하기

1~n번까지 서로 다른 번호를 가진 온실 `n`개가 있습니다. 각 온실은 식물을 키우는 ‘생장력’을 갖고 있으며, 연속된 번호의 온실들을 합쳐 그룹을 만들 수 있습니다.  
이렇게 형성된 그룹의 총 생장력은 **해당 그룹에 속한 온실들의 생장력 합 + 과거에 최적화된 그룹 수 × 2배 보너스**로 계산됩니다.

**최적화된 온실 그룹**은 이미 과거에 프로젝트를 통해 최적화된 그룹이며, 반드시 연속된 번호의 온실로 구성되어 있습니다.  
최적화된 그룹에 포함된 온실은 다른 최적화 그룹에 포함될 수 없습니다.

---

### ✅ 목표

당신은 **총 생장력 `m` 이상을 만족하는 온실 그룹 조합의 개수**를 구하려고 합니다.

---

### 📥 입력

- `growths`: 각 온실의 생장력을 나타내는 정수 배열 (길이 `n`)
- `pasts`: 과거 최적화된 온실 그룹의 범위를 담은 `[start, end]` 쌍의 배열
- `m`: 필요한 최소 생장력

---

### 🔄 출력

- 총 생장력이 `m` 이상이 되는 **온실 그룹 조합의 개수**를 정수로 반환하세요.

---

### 🔒 제한사항
- 1 ≤ `growths`의 길이 = `n` ≤ 300,000
  - `growths[i]`는 i+1번 온실의 생장력입니다.
- 0 ≤ `pasts`의 길이 ≤ 300,000
  - 각 요소는 `[start, end]` 형태로 1 ≤ start ≤ end ≤ n 범위를 가집니다.
- 1 ≤ `m` ≤ 10^15

---

#### 입출력 예
| growths         | pasts              | m  | result |
|----------------|--------------------|----|--------|
| [1, 2, 3, 4, 5] | [[1, 2], [4, 5]]    | 10 | 2      |
| [1, 1, 1, 1, 1] | [[2, 3]]           | 3  | 6      |
| [10, 1, 1, 1, 10]| []                | 11 | 3      |
