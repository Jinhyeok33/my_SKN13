# 🎟️ 기간 내 책 읽은 수 만큼 당첨되는 이벤트 LV.??

### 🎯 문제 설명  
여러 개의 책 읽기 이벤트가 진행됩니다.  
각 이벤트는 다음과 같은 정보를 가집니다:  
- 이벤트 기간 (`start`일부터 `end`일까지)  
- 해당 기간에 채워야 할 **최소 리뷰 수**  

모든 이벤트에 대해 **공통 최대 당첨자 수** `m`이 주어집니다.  

각 참가자는 특정 날짜에 책을 읽은 기록이 있으며, 해당 날짜가 이벤트 기간 내에 속할 경우 그 리뷰가 해당 이벤트에 반영됩니다.

---

## 📋 이벤트 당첨 규칙  
1. 각 이벤트마다 기간 내 누적 리뷰 수가 `min_reviews` 이상인 참가자를 **후보**로 판단  
2. 후보 수가 `m` 이하라면 **모두 당첨**  
3. 후보 수가 `m`를 초과하면, **최초로 `min_reviews`에 도달한 순서**대로 앞에서부터 `m`명만 당첨  

---

## 🔒 제한사항  
- `events[i]`는 `[start_day, end_day, min_reviews]` 형식입니다.  
- 공통 최대 당첨자 수: 정수 `m`  
- `reviews[i]`는 `"참가자명 날짜 권수"` 형식입니다. (예: `"arin 20 2"`)  
- 참가자 이름은 영문 소문자, 최대 10명 이하  
- 날짜는 1 이상 31 이하의 정수  
- 권수는 1 이상의 정수  
- 이벤트 개수 최대 50개, 리뷰 개수 최대 1000건

사전순으로 정렬된 참가자 이름에 대응하여, 각 참가자가 **총 몇 번 당첨**되었는지를 담은 1차원 정수 배열을 반환합니다.

| events                  | m  | reviews                                                                                      | result    |
